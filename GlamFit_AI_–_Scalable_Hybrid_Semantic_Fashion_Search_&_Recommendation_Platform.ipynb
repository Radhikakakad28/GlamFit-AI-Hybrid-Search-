```python
import numpy as np
import faiss
from sentence_transformers import SentenceTransformer
from rank_bm25 import BM25Okapi


```

```python
df.columns
```

```python
df["search_text"] = (
    df["product_name"].astype(str) + " " +
    df["product_type"].astype(str) + " " +
    df["color"].astype(str) + " " +
    df["section"].astype(str)
)

df.head()
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
embeddings = model.encode(df["search_text"].tolist(), show_progress_bar=True)

dimension = embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
index.add(np.array(embeddings))
```

```python
tokenized_corpus = [doc.split() for doc in df["search_text"]]
bm25 = BM25Okapi(tokenized_corpus)
```

```python
def hybrid_search(query, top_k=5, alpha=0.6):

    # Semantic Search
    query_embedding = model.encode([query])
    distances, indices = index.search(np.array(query_embedding), top_k)
    semantic_scores = 1 / (1 + distances[0])

    # BM25
    tokenized_query = query.split()
    bm25_scores = bm25.get_scores(tokenized_query)
    bm25_scores = bm25_scores / np.max(bm25_scores)

    # Combine
    results = []
    for idx_pos, idx in enumerate(indices[0]):
        score = alpha * semantic_scores[idx_pos] + (1-alpha) * bm25_scores[idx]
        results.append((idx, score))

    results = sorted(results, key=lambda x: x[1], reverse=True)

    return df.iloc[[r[0] for r in results]]
```

```python
hybrid_search("red party dress for women")
```

```python

user_profile = {
    "preferred_color": "Red",
    "preferred_section": "Women",
    "preferred_type": "Dress"
}
```

```python
def personalize_results(results, user_profile):

    personalized = results.copy()
    personalized["personal_score"] = 0

    for index, row in personalized.iterrows():

        score = 0

        if user_profile["preferred_color"].lower() in row["color"].lower():
            score += 1

        if user_profile["preferred_section"].lower() in row["section"].lower():
            score += 1

        if user_profile["preferred_type"].lower() in row["product_type"].lower():
            score += 1

        personalized.at[index, "personal_score"] = score

    personalized = personalized.sort_values(by="personal_score", ascending=False)

    return personalized
```

```python
results = hybrid_search("party dress")
personalized_results = personalize_results(results, user_profile)

personalized_results[["product_name","color","section","personal_score"]]
```

```python
import random

reviews = [
    "Amazing quality and very comfortable",
    "Poor stitching and bad material",
    "Perfect for party wear, loved it",
    "Not worth the price",
    "Very stylish and elegant design",
    "Color faded after one wash"
]

df["review_text"] = [random.choice(reviews) for _ in range(len(df))]
```

```python
!pip install transformers
```

```python
from transformers import pipeline

sentiment_model = pipeline("sentiment-analysis")
```

```python
df["sentiment"] = df["review_text"].apply(lambda x: sentiment_model(x)[0]["label"])
df[["product_name","review_text","sentiment"]].head()
```

```python
df["is_fake"] = df["review_text"].apply(lambda x: 1 if "price" in x.lower() else 0)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df["review_text"])
y = df["is_fake"]

model_fake = LogisticRegression()
model_fake.fit(X, y)
```

```python
sample_review = ["This product is amazing and premium quality"]
sample_vec = vectorizer.transform(sample_review)

model_fake.predict(sample_vec)
```
